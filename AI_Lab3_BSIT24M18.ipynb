```python
list1 = []

n = int(input("How many numbers do you want to enter? "))

for i in range(n):
    number = int(input("Enter number: "))
    list1.append(number)

even = 0
odd = 0

for i in list1:
    if i % 2 == 0:
        even +=1
    else:
        odd +=1

print("Number of even numbers:", even)
print("Number of odd values:", odd)

# without taking input
list1= [1,3,12,14,35,24]
even=0
odd=0
for i in list1:
   if(i%2==0):
    even+=1
   else:
      odd +=1
print("Number of even numbers",even,"\n Number of odd values",odd)
```

```python
rows = int(input("Enter the number of rows: "))
cols = int(input("Enter the number of columns: "))

arr = []

for i in range(rows):
    row = []

    for j in range(cols):
        row.append(i * j)

    arr.append(row)

for row in arr:
    print(row)
```

```python
string=input("Enter a sequence of 4 digit binary nums separated by commas: ")
list=string.split(",")
print("OUTPUT: ")
for binary_string in list:
    decimal=int(binary_string,2)
    if decimal%5==0:
        print(f"{binary_string} ({decimal})")
```

```python
string=input("Enter a string and i will tell you the number of letters and digits it contains: ")
digits=0
letters=0
for char in string:
  if char.isdigit():
    digits += 1
  if char.isalpha():
    letters += 1
  print(f"Numbers of letters: {letters}")
  print(f"Numbers of digits: {digits}")
```
