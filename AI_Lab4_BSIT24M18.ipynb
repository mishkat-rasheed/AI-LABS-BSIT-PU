```python
# Reversing a string using stack it is same for queue also
string=input("Enter the string : ")
stack = []
for i in string:
    stack.append(i)
reversed_string = ""
while stack:
    reversed_string += stack.pop()
print("Reversed string:", reversed_string)
```

```python
arr = [10, 20, 30, 40, 50, 60, 70]

target = int(input("Enter number to search: "))

low = 0
high = len(arr) - 1
found = False

while low <= high:

    mid = (low + high) // 2  # Floor division function

    if arr[mid] == target:
        print("Element found at index:", mid)
        found = True
        break

    elif target > arr[mid]:
        low = mid + 1

    else:
        high = mid - 1

if found == False:
    print("Element not found")
```
